```python
# The BNS data without GW constraints
df = pd.read_csv('./MCL_BNS_new.dat',delim_whitespace=True)
#df = df.sort_values(by=[‘mass’])

mass = df['mass'].tolist()

tides = df['tides'].tolist()
prob = df['prob'].tolist()
prob = np.array(prob)
tides = np.array(tides)
mass = np.array(mass)
```

```python
#plt.figure()
#plt.scatter(tides,mass,c=np.log10(prob),cmap='Purples',alpha=1.0)
#plt.scatter(tides,mass,c=np.log10(prob),cmap='Blues',alpha=1.0)
plt.scatter(tides,mass,c=np.log10(prob),cmap='GnBu',alpha=1.0)
#plt.axhspan(1.186, 1.192, color='k', alpha=0.4)
#plt.annotate(r'GW170817', (800,1.212), fontsize=20)

# GW190425
# plt.axhline(1.44, c='k', ls='-.')
plt.axhspan(1.42, 1.46, color='red', alpha=0.4)
plt.annotate(r'GW190425', (775,1.48), fontsize=20)
#plt.colorbar(label=r'$\rm \log_{10}[\pi(\mathcal{M},\,\tilde{\Lambda})]$')
cb = plt.colorbar()
cb.set_label(label=r'$\rm \log_{10}[\pi(\mathcal{M},\,\tilde{\Lambda})]$', fontsize=20)
# GW170817
# plt.axhline(1.186, c='k', ls='--')
plt.axhspan(1.186, 1.192, color='red', alpha=0.4)
plt.annotate(r'GW170817', (775,1.212), fontsize=20)
plt.xlim(0, 1000)
plt.ylim(min(mass), max(mass))
print(min(mass))
print(max(mass))
plt.ylabel(r'$\mathcal{M}\,\left[M_\odot\right]$', fontsize=20)
plt.xlabel(r'$\tilde{\Lambda}$', fontsize=20)
#ax1.set_aspect('equal', adjustable='box')
#ax2.set_aspect('equal', adjustable='box')
#figure.tight_layout()
#plt.savefig('BNS_chirpmass_prior_noGW.png',dpi=500,bbox_inches='tight')
plt.savefig('BNS_chirpmass_prior_noGW.pdf',dpi=500,bbox_inches='tight')
```
